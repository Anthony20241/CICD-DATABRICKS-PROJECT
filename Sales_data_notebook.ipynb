```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = SparkSession.builder.getOrCreate()

# Sample data
data = [
    ("2025-01-01", "North", "Laptop", 1200, 3),
    ("2025-01-01", "South", "Mouse", 25, 20),
    ("2025-01-02", "West", "Keyboard", 45, 12),
    ("2025-01-02", "North", "Monitor", 200, 5),
    ("2025-01-03", "East", "Laptop", 1300, 2),
    ("2025-01-03", "South", "Headset", 80, 10)
]

columns = ["date", "region", "product", "unit_price", "quantity"]

df_sales = spark.createDataFrame(data, columns)

df_sales.show()
```

```python
#Sales analyss

df_sales = df_sales.withColumn("total_sales", col("unit_price") * col("quantity"))
df_sales.show()
```

```python
from pyspark.sql.functions import sum as _sum

df_sales.groupBy("region") \
    .agg(_sum("total_sales").alias("total_sales")) \
    .orderBy("total_sales", ascending=False) \
    .show()
```

```python
from pyspark.sql.functions import avg

df_sales.groupBy("region") \
    .agg(avg("total_sales").alias("avg_sales")) \
    .orderBy("avg_sales", ascending=False)
```
